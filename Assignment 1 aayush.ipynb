```python
# Student Data Management using Dictionary, Tuple, and List
# Done by: Aayush

# Dictionary of student records
# Key -> Roll Number, Value -> Tuple(Name, Branch, Marks List)
students = {
    101: ("Aarav", "Computer Science", [85, 90, 78]),
    102: ("Priya", "Mechanical", [70, 65, 80]),
    103: ("Rohan", "Electronics", [88, 92, 95]),
    104: ("Aayush", "Computer Science", [91, 87, 94])
}

print("Initial Student Records:")
for roll, info in students.items():
    print(roll, info)

# 1. Add a new student record
students[105] = ("Sneha", "Civil", [75, 82, 79])
print("Added a new student (Roll No 105 - Sneha)")

# 2. Update an existing student's details
students[102] = ("Priya", "Electrical", [90, 85, 88])
print("Updated Roll No 102 (Priya) - Branch and Marks changed")

# 3. Delete an existing student record
del students[101]
print("Deleted Roll No 101 (Aarav)")

# 4. Display final records
print("Final Student Records after Add, Update, and Delete:")
for roll, info in students.items():
    print(roll, info)

print("All operations completed successfully.")
```
